```python
import seaborn as sns
```

```python
grafico = sns.histplot(data=df, x="categoria",kde=True, bins=5)
grafico.set_title ("vendas por categoria")
grafico.set_xlabel("categoria")
grafico.set_ylabel("vendas")
```

```python
grafico= sns.boxplot(data=df,x="categoria",y="preco_unitario")
grafico.set_title ("preço por categoria")
grafico.set_xlabel("categoria")
grafico.set_ylabel("preços")
```
